# Non-primitive data types:
1) List
2) Tuple
3) Set
4) Dictionary
5) String
6) Frozen Set

```python
# List Operations:
# CREATION:
a = list((1,2,4))
print(type(a))
```

```python
# READ:
print(a)
print(linespace)
for i in a:
    print(i)
print(linespace)
print(a[-2])
```

```python
# UPDATE:
a.append(2)
print(a)
print(linespace)
a.insert(0,10)
print(a)

a.extend([1,7,9])
print(a)
```

## DELETE
1) pop()
2) remove()
3) clear()
4) del

```python
# DELETE:
print(a)
print(linespace)
a.pop()
print(a)
```

```python
a.pop(2)
print(a)
```

```python
a.remove(100)
print(a)
```

```python
a.clear()
print(a)

del a
print(a)
```
